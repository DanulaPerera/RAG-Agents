# Document Loaders (Langchain)

```python
from langchain_community.document_loaders import PyPDFLoader, TextLoader
```

```python
# Load a PDF document
pdf_loader = PyPDFLoader("../data/2507.03347v1.pdf", mode="single")
pdf_docs = pdf_loader.load()
```

```python
len(pdf_docs)
```

```python
pdf_docs
```

```python
# Load a Text document
txt_loader = TextLoader("../data/king-lear_TXT_FolgerShakespeare.txt")  # TextLoader works for .txt or .md
txt_docs = txt_loader.load()
```

```python
txt_docs
```

# Text Splitting

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100,
)
```

```python
my_docs = pdf_docs + txt_docs
```

```python
my_docs
```

```python
texts = text_splitter.split_documents(my_docs)
```

```python
texts
```

```python
len(texts)
```

# Vector Stores

```python
import os
from langchain_pinecone import PineconeVectorStore
from pinecone import Pinecone
from langchain_openai import OpenAIEmbeddings
```

```python
pc = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))
index = pc.Index("stemlink-ai-bootcamp-demo")
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

vector_store = PineconeVectorStore(embedding=embeddings, index=index)
```

```python
from uuid import uuid4

document_ids = [str(uuid4()) for _ in range(len(texts))]
```

```python
len(document_ids)
```

```python
vector_store.add_documents(documents=texts, ids=document_ids)
```

# Embeddings

# Retrieval

```python
query = "What are the results of BBH benchmark on unstructured reasoning?" 
results = vector_store.similarity_search(query, k=4)
```

```python
results
```

```python
query = "What is the shakespere story about?" 
results = vector_store.similarity_search(query, k=4)
```

```python
query = "How does structured reasoning work?"  
results = vector_store.similarity_search_with_score(query, k=4)
```
