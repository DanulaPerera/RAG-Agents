# Document Loaders (LangChain)

```python
from langchain_community.document_loaders import PyPDFLoader,TextLoader
```

```python
file_path_pdf = "../data/2507.03347v1.pdf"
loader = PyPDFLoader(file_path_pdf, mode="single")
data = loader.load()

print(data)

```

```python
data
```

```python
file_path_md = "../data/king-lear_TXT_FolgerShakespeare.txt"
text_loader = TextLoader(file_path_md)
text_docs = text_loader.load()

text_docs
```

```python
print(text_docs[0].page_content)
```

# Text Splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100
)
```

```python
my_docs = data + text_docs
```

```python
my_docs
```

```python
splitted_texts = text_splitter.split_documents(my_docs)
```

```python
splitted_texts
```

# Vector Stores

```python
import os
from langchain_pinecone import PineconeVectorStore
from pinecone import Pinecone
from langchain_openai import OpenAIEmbeddings
```

```python
import pinecone

pc = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))
index = pc.Index("rag-agent-1-stemlink")
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
vector_store = PineconeVectorStore(embedding=embeddings,index=index)
```

```python
from uuid import uuid4

document_ids = [str(uuid4()) for _ in range(len(splitted_texts))]
```

```python
len(document_ids)
```

```python
vector_store.add_documents(documents=splitted_texts, ids=document_ids )
```

# Retrievel

```python
query = "What are the Results BBH benchmark of Unstructured reasoning?"
results = vector_store.similarity_search_with_score(query, k=4)
```

```python
results
```
